# Creating Agents With Langchain

### AI Agent with LangChain – Introduction

An AI Agent is an intelligent system that can understand user inputs, reason about tasks, make decisions, and take actions to achieve specific goals. Unlike traditional chatbots, AI agents can interact with external tools, APIs, databases, and other resources to solve complex problems.

- **LangChain** is a powerful framework that simplifies the development of AI agents by connecting Large Language Models (LLMs) with tools, memory, and external data sources. It provides components for prompt management, reasoning, retrieval, tool usage, and workflow orchestration.

- Using LangChain, developers can build context-aware AI agents that can search the web, query databases, analyze documents, maintain conversation history, and automate tasks. This makes it easier to create intelligent applications that can understand, reason, and act in real-world scenarios.

- In simple terms, LangChain acts as the bridge between an AI model and the external world, enabling the creation of smarter, more capable, and interactive AI systems.


```python
import os
from dotenv import load_dotenv
load_dotenv()

os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
```

```python
from langchain.agents import create_agent
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="qwen3:8b"
)

def get_weather(city:str)->str:
    """Get the weather for a city."""
    return f"The weather in {city} is sunny."

agent = create_agent(
    model =llm,
    tools=[get_weather],
    system_prompt = "you are a helpful assistant."
)
agent
```

```python
## run the agent

agent.invoke({"messages":[{"role":"user", "content":"what is the weather in new york"}]})
```

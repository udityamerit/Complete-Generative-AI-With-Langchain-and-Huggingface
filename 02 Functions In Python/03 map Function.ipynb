# Map Function

```python
numbers = [1,2,3,4,5,5]

squared = list(map(lambda x: x*x, numbers))

print(squared)
```

```python
def square(x):
        return x*x

list(map(square, numbers))
```

```python
n1 = [1,2,3,4,5]
n2 = [5,4,3,2,1]

added = list(map(lambda x, y: x+y, n1,n2))
```

```python
l = ["hello", "hi", "what are you doing"]

upper = list(map(lambda x: x.upper(), l))

print(upper)
```

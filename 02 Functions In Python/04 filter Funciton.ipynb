# filter() function

```python
def even(l):
        if l%2==0:
                return True

even(10)
```

```python
lst = [1,2,3,4,5,6,7,8,9]

list(filter(even, lst))
```

```python
greater_than_five = list(filter(lambda x: x>5, lst))
print(greater_than_five)
```

```python
greater_than_five_even = list(filter(lambda x: x>5 and x%2==0, lst))
print(greater_than_five_even)
```

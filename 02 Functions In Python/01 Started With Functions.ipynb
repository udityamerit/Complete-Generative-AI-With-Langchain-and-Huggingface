# Started with Function

```python
# Creating a function

def even_or_odd(num):
        if num%2==0:
                print(f"{num} is even number")

        else:
                print(f"{num} is odd number")
                
```

```python
# calling the function

number=23

even_or_odd(number)
```

```python
# function with multiple parameters

def mul_par(a, b, c, d):
        ans = a+b*c/d
        return ans

call = mul_par(1,2,3,4)

print(call)
```

```python
# function with default parameters

def greet(name="Guest"):
        print(f"Hello {name} Welcome to this new generative ai course")

greet()
```

```python
greet("uditya")
```

### variable length arguments 

```python
## positional and keywords arguments

def positional_args(*args):
        for arguments in args:
                print(arguments, end=" ")


positional_args(1,2,3,4,5,"uditya", 'a', 'b')
```

```python
### keywords arguments

def key_args(**kwargs):
        for key, value in kwargs.items():
                print(f"{key}: {value}")

key_args(name="Uditya", age=23, city="Jaunpur", country="India")
```

```python
def poaug_kwarg(*args, **kwargs):
        for val in args:
                print(f"positional argument: {val}")

        for key, value in kwargs.items():
                print(f"kwargs: {key}: {value}")


poaug_kwarg(1,2,3,4,5, "Udityamerit", name="Udityamerit", age=45, education="B.Tech", stream = "CSE (AI&ML)")
```

```python
def pol(param):
        param = param.lower().replace(" ", "")
        i = 0
        n = len(param)-1
        while(i<n):
                if param[i]==param[n]:
                        i+=1
                        n-=1
                        continue
                else:
                        return "Not a polindrom"     
                        
        return "Polindrom"


print(pol("hel leh"))
```

```python
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_huggingface import HuggingFaceEndpoint
```

```python
import os
from langchain_core.messages import HumanMessage
from langchain_huggingface import HuggingFaceEndpoint, ChatHuggingFace

repo_id = "deepseek-ai/DeepSeek-R1"

llm = HuggingFaceEndpoint(
    repo_id=repo_id,
    task="conversational",  
    huggingfacehub_api_token=os.getenv("HF_TOKEN"),
    max_new_tokens=150, 
    temperature=0.7
)

chat_model = ChatHuggingFace(llm=llm)

messages = [
    HumanMessage(content="What is genai")
]

response = chat_model.invoke(messages)
print(response.content)
```

```python
from langchain_classic.prompts import PromptTemplate
template = """
Question:{question}
Answer: Lets think step by step.
"""
prompt = PromptTemplate(template=template, input_variables=["question"])
print(prompt)
```

```python
llm_chain = prompt | llm
```

```python
response = llm_chain.invoke(
    {"question": "Who won the cricket world cup of 2026?"}
)

print(response)
```

```python
from langchain_huggingface import HuggingFaceEmbeddings

embeddings = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-mpnet-base-v2",
    encode_kwargs={"normalize_embeddings": True},
)

query_result = embeddings.embed_query("This is a test document.")
doc_result = embeddings.embed_documents(["This is a test document."])
```

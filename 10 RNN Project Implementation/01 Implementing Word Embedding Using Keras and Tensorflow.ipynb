# Implementing Word Embedding Using Keras and Tensorflow

```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
## Sentence

sent = [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am good developer',
    'understand the meaning of words',
    'your videos are good',
]
```

```python
sent
```

```python
voc_size = 10000 ## defining the vocabulary size
```

```python
## One hot representation

one_hot_repr = [one_hot(words, voc_size) for words in sent]
one_hot_repr
```

```python
## word Embedding Representation

from tensorflow.keras.layers import Dense, Embedding, Input
from tensorflow.keras.utils import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

```python
sent_length = 8
embedded_doc = pad_sequences(one_hot_repr, padding='post', maxlen=sent_length)
print(embedded_doc)
```

```python
dim = 10 # setting the dimension
```

```python
model = Sequential()
model.add(Input(shape=(sent_length,)))
model.add(Embedding(voc_size, dim))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
emb = model.predict(embedded_doc)
```

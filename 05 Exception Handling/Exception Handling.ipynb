# Exception Handling

```python
try:
        a=b
except:
        print("The variable has not been assign")
```

```python
try:
        a=b
except NameError as ex:
        print(ex)
```

```python
try:
        result = 1/2
        a=b 
except ZeroDivisionError as ex:
        print(ex)

except Exception as ex1:
        print(ex1)
        
```

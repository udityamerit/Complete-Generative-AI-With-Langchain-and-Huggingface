# Introduciton To Chroma DB

```python
from langchain_chroma import Chroma
```

```python
from langchain_community.document_loaders import TextLoader
from langchain_ollama import OllamaEmbeddings
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader('test.txt')
data = loader.load()
data
```

```python
text_splitter =  RecursiveCharacterTextSplitter(chunk_size = 1000, chunk_overlap = 0)
split = text_splitter.split_documents(data)
```

```python
embeddings = OllamaEmbeddings(model = 'nomic-embed-text')
```

```python
vector_db = Chroma.from_documents(documents= split, embedding=embeddings)
```

```python
query = "what is my project names?"
docs = vector_db.similarity_search(query=query)
```

```python
docs[0].page_content
```

```python
vector_db = Chroma.from_documents(documents= split, embedding=embeddings, persist_directory='./chroma_db')
```

```python
db2 = Chroma(persist_directory='./chroma_db', embedding_function=embeddings)
docs = db2.similarity_search(query)
```

```python
docs[1].page_content
```

```python
docs = vector_db.similarity_search_with_score(query)
docs
```

```python
retriver = vector_db.as_retriever()
retriver.invoke(query)[0]
```

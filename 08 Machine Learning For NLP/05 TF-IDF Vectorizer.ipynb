```python
import pandas as pd
```

```python
try:
    data = pd.read_csv(r"..\01 Datasets\spam.csv", encoding="utf-8", names=["v1", "v2"])
except UnicodeDecodeError:
    data = pd.read_csv(r"..\01 Datasets\spam.csv", encoding="latin-1")
```

```python
data
```

```python
data[["label", "message"]] = data[["v1", "v2"]]
data = data[["label", "message"]]
```

```python
data.head()
```

```python
import nltk
import re

nltk.download("stopwords")
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer

wordlemmatize = WordNetLemmatizer()
```

```python
corpus = []
for i in range(len(data)):
    review = re.sub("[^a-zA-Z]", " ", data["message"][i])
    review = review.lower()
    review = review.split()
    review = [
        wordlemmatize.lemmatize(word)
        for word in review
        if not word in stopwords.words("english")
    ]
    review = " ".join(review)
    corpus.append(review)
```

# Create TF-IDF And N-Grams

```python
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf = TfidfVectorizer(max_features=100)
x = tfidf.fit_transform(corpus).toarray()
```

```python
import numpy as np

np.set_printoptions(
    edgeitems=30, linewidth=100000, formatter=dict(float=lambda x: "%.3g" % x)
)
```

# N-Grams

```python
tfidf = TfidfVectorizer(max_features=100, ngram_range=(2, 2))
x = tfidf.fit_transform(corpus).toarray()
```

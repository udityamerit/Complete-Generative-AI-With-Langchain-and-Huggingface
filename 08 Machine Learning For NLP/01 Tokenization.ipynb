```python
corpus = """ In Natural Language Processing (NLP), a corpus acts as the foundational dataset that teaches machines how humans communicate. It is a massive, structured collection of written texts or spoken audio files carefully gathered from sources like books, websites, news articles, or social media. Instead of relying on rigid, pre-programmed grammar rules, modern AI models analyze these corpora using statistical methods and machine learning to independently discover patterns, definitions, and context. Depending on the project, a corpus can be general—covering everyday language—or highly specialized to focus on specific domains like medicine, law, or finance. Engineers frequently use annotated corpora, where the text is manually labeled with tags like parts of speech or emotional sentiment, to train models for complex tasks such as language translation, chatbots, and automated text summarization.
"""
```

```python
print(corpus)
```

```python
import nltk

nltk.download("punkt_tab")
```

## Tokenization
## Sentence--> paragraph

```python
from nltk.tokenize import sent_tokenize

doc = sent_tokenize(corpus)
```

```python
doc
```

```python
for sentence in doc:
    print(sentence)
    print(" ")
```

## Paragraph --> words
## Sentence --> words

```python
from nltk.tokenize import word_tokenize

word_doc = word_tokenize(corpus)
```

```python
word_doc
```

```python
from nltk.tokenize import wordpunct_tokenize

wordpunct_tokenize(corpus)
```

```python
from nltk.tokenize import TreebankWordTokenizer

tokenizer = TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
```

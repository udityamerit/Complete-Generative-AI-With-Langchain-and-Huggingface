```python
para = """
            Dr. A.P.J. Abdul Kalam’s speeches were powerful, deeply inspiring, and universally focused on empowering the youth to transform India into a developed nation. He consistently urged students to dream big, famously stating that dreams transform into thoughts, which ultimately result in action. To achieve these dreams, he championed a distinct four-step success formula: setting a great aim, continuously acquiring knowledge, working hard, and maintaining the perseverance to overcome inevitable failures. He believed that true leadership is defined by defeating problems rather than letting problems defeat you. Furthermore, Dr. Kalam tied personal character directly to global peace, teaching that righteousness in the individual heart creates beauty in character, harmony in the home, order in the nation, and ultimately peace in the world.
"""
```

```python
import nltk

sentences = nltk.sent_tokenize(para)
```

```python
sentences
```

```python
from nltk.corpus import stopwords
import nltk

nltk.download("averaged_perceptron_tagger_eng")
```

```python
from prettytable import PrettyTable

for i in range(len(sentences)):
    words = nltk.word_tokenize(sentences[i])
    words = [word for word in words if word not in set(stopwords.words("english"))]
    pos_tags = nltk.pos_tag(words)

    # 1. Initialize the PrettyTable
    table = PrettyTable()
    table.field_names = ["Word", "POS Tag"]

    # 2. Populate rows with the token and its tag
    for word, tag in pos_tags:
        table.add_row([word, tag])
    print(table)
```

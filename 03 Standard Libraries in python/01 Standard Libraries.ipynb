```python
## data serialization

import json

data = {
        'name':'Uditya',
        'age':25
}

json_str=json.dumps(data)
print(json_str)
print(type(json_str))

parsed_data = json.loads(json_str)
print(parsed_data)
print(type(parsed_data))
```

```python
# writing the csv file
import csv

with open('../01 Datasets/test.csv', mode = 'w', newline='') as file:
        writer=csv.writer(file)
        writer.writerow(['name', 'age', 'course', 'job'])
        writer.writerow(['Uditya', 23, 'CSE', 'meta'])
        writer.writerow(['Rutu', 21, 'CSE(AL/ML)', 'amazon'])
        writer.writerow(['Sai', 20, 'CSE', 'google'])
```

```python
# read the csv

with open( '../01 Datasets/test.csv', mode='r') as file:
        reader = csv.reader(file)
        for row in reader:
                print(row)
```
